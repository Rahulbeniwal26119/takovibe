## 💼 Real-World Example: Debug a Mysterious Bug

<div style="background: #fff3cd; border: 1px solid #ffeaa7; padding: 20px; border-radius: 10px; margin: 20px 0;">
    <div style="display: flex; align-items: center;">
        <span style="font-size: 2em; margin-right: 15px;">🕵️‍♂️</span>
        <div>
            <h4 style="margin: 0; color: #856404;">The Mystery</h4>
            <p style="margin: 5px 0 0 0; color: #856404;">You have a function that processes user data, but it's giving wrong results for some inputs. Let's solve it together!</p>
        </div>
    </div>
</div>


```python
import ipdb
from rich.console import Console
from rich.table import Table

console = Console()


def process_user_scores(users_data):
    """
    Process user scores and calculate statistics.
    This function has a mysterious bug - can you find it with ipdb?
    """
    console.print("🔍 [bold blue]Processing user scores...[/bold blue]")

    # Initialize variables
    total_score = 0
    user_count = 0
    processed_users = []

    for user in users_data:
        # 🔥 UNCOMMENT THE NEXT LINE TO START DEBUGGING!
        # ipdb.set_trace()  # <-- Your debugging superpower starts here!

        name = user.get("name", "Unknown")
        scores = user.get("scores", [])

        if scores:  # Only process users with scores
            user_average = sum(scores) / len(scores)
            total_score += user_average  # 🐛 Suspicious line - is this right?
            user_count += 1

            processed_users.append(
                {
                    "name": name,
                    "average": round(user_average, 2),
                    "total_points": sum(scores),
                    "num_tests": len(scores),
                }
            )

    # Calculate overall statistics
    if user_count > 0:
        global_average = total_score / user_count
    else:
        global_average = 0

    return {
        "users": processed_users,
        "global_average": round(global_average, 2),
        "total_users": user_count,
    }


# 🧪 Test data - some users have different numbers of scores
test_users = [
    {"name": "Alice", "scores": [85, 92, 78, 96]},  # 4 tests
    {"name": "Bob", "scores": [75, 82]},  # 2 tests
    {"name": "Charlie", "scores": [95, 88, 92, 90, 87]},  # 5 tests
    {"name": "Diana", "scores": []},  # No tests
    {"name": "Eve", "scores": [70, 75, 80]},  # 3 tests
]

console.print("\n🎯 [bold green]Testing the function with sample data:[/bold green]")
console.print("=" * 60)

# Run the function
result = process_user_scores(test_users)

# Display results in a beautiful table
table = Table(title="📊 User Score Analysis Results")
table.add_column("User", style="cyan", no_wrap=True)
table.add_column("Average", style="magenta")
table.add_column("Total Points", style="green")
table.add_column("Tests Taken", style="yellow")

for user in result["users"]:
    table.add_row(
        user["name"],
        str(user["average"]),
        str(user["total_points"]),
        str(user["num_tests"]),
    )

console.print(table)
console.print(f"\n🏆 [bold]Global Average: {result['global_average']}[/bold]")
console.print(f"👥 [bold]Total Users Processed: {result['total_users']}[/bold]")

print("\n" + "=" * 60)
print("🤔 QUESTION: Does the global average look correct to you?")
print("💡 TIP: Uncomment the ipdb.set_trace() line above and run again!")
print("🔍 Use ipdb to inspect variables and find the bug!")
print("=" * 60)
```

### 🎯 **Your Mission (Should You Choose to Accept It)**

<div style="background: #fff3cd; border: 1px solid #ffeaa7; padding: 20px; border-radius: 10px; margin: 15px 0;">
    <h4 style="margin-top: 0; color: #856404;">🕵️‍♂️ Detective Work Required</h4>
    <p style="color: #856404;">The function below contains <strong>3 deliberate bugs</strong>. Use ipdb to find and understand each one!</p>
    <div style="background: #f8f9fa; padding: 15px; border-radius: 5px; margin: 10px 0;">
        <p style="margin: 0; color: #495057;"><strong>🎯 Goals:</strong></p>
        <ol style="color: #495057; margin: 10px 0 0 20px;">
            <li>Uncomment the ipdb line in the code below</li>
            <li>Run the cell and use ipdb commands to explore</li>
            <li>Find all 3 bugs using ipdb investigation</li>
            <li>Understand why each bug occurs</li>
        </ol>
    </div>
</div>


```python
import ipdb
from rich.console import Console
from rich.panel import Panel

console = Console()


def calculate_team_performance(team_data):
    """
    Calculate team performance metrics.
    🐛 This function contains 3 bugs - find them with ipdb!

    Expected behavior:
    - Calculate individual scores
    - Find team average
    - Identify top performer
    - Handle edge cases gracefully
    """
    console.print("🔥 [bold blue]Calculating team performance...[/bold blue]")

    # 🚨 UNCOMMENT THE NEXT LINE TO START YOUR DETECTIVE WORK!
    # ipdb.set_trace()  # <-- Your debugging journey starts here!

    total_score = 0
    member_count = 0
    individual_scores = []

    for member in team_data:
        name = member["name"]
        tasks_completed = member["tasks_completed"]
        hours_worked = member["hours_worked"]

        # Bug #1: Mathematical error in score calculation
        if hours_worked > 0:
            efficiency = tasks_completed / hours_worked
            # This should be a more complex calculation, but there's a bug!
            score = efficiency * 100 + tasks_completed  # 🐛 Bug lurking here
        else:
            score = 0

        individual_scores.append(
            {
                "name": name,
                "score": score,
                "efficiency": efficiency if hours_worked > 0 else 0,
            }
        )

        total_score += score
        member_count += 1

    # Bug #2: Division by zero potential
    team_average = total_score / member_count  # 🐛 What if member_count is 0?

    # Bug #3: Incorrect logic for finding top performer
    top_performer = individual_scores[0]  # 🐛 Assumes list has items and isn't sorted
    for member in individual_scores:
        if member["score"] < top_performer["score"]:  # 🐛 Wrong comparison operator!
            top_performer = member

    return {
        "team_average": round(team_average, 2),
        "top_performer": top_performer,
        "individual_scores": individual_scores,
        "total_members": member_count,
    }


# 🧪 Test data with various edge cases
test_team = [
    {"name": "Alice", "tasks_completed": 15, "hours_worked": 40},
    {"name": "Bob", "tasks_completed": 12, "hours_worked": 35},
    {"name": "Charlie", "tasks_completed": 8, "hours_worked": 30},
    {"name": "Diana", "tasks_completed": 20, "hours_worked": 45},
    # Uncomment this to test edge case:
    # {'name': 'Eve', 'tasks_completed': 5, 'hours_worked': 0},  # Zero hours!
]

console.print("\n" + "=" * 60)
console.print("🎯 [bold green]IPDB DEBUGGING CHALLENGE[/bold green]")
console.print("=" * 60)

try:
    result = calculate_team_performance(test_team)

    # Display results
    console.print(
        Panel.fit(
            f"🏆 Team Average: {result['team_average']}\n"
            f"🌟 Top Performer: {result['top_performer']['name']} "
            f"(Score: {result['top_performer']['score']})\n"
            f"👥 Total Members: {result['total_members']}",
            title="📊 Team Performance Results",
            border_style="green",
        )
    )

    # Show individual scores
    console.print("\n📋 [bold]Individual Scores:[/bold]")
    for member in result["individual_scores"]:
        console.print(f"  • {member['name']}: {member['score']:.2f} points")

except Exception as e:
    console.print(f"💥 [bold red]Error occurred: {e}[/bold red]")
    console.print("🔍 This might be one of the bugs! Use ipdb to investigate!")

console.print("\n" + "=" * 60)
console.print("🤔 [bold yellow]DETECTIVE QUESTIONS:[/bold yellow]")
console.print("1. 🔍 Does the top performer look correct?")
console.print("2. 🧮 Are the score calculations making sense?")
console.print("3. 🛡️ What happens with edge cases (zero hours)?")
console.print("4. 📊 Is the team average calculation robust?")
console.print(
    "\n💡 [bold]TIP:[/bold] Use ipdb commands like 'p', 'pp', 'n', 's' to investigate!"
)
console.print("=" * 60)
```
